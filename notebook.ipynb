```python
# Voy a hacer un nuevo df con las métricas que identifican eficiencia, creando un df para cada métrica y luego unirlas con merge en operator_id único
# 1. Llamadas pérdidas entrantes y llamadas entrantes con mucho tiempo de espera.
#Llamadas entrantes
inbound_calls = dataset_operators[dataset_operators['direction']=='in']
# total de llamadas entrantes
total_inbound_calls_per_operator = inbound_calls.groupby('operator_id').size()
#print(inbound_calls)
#print(total_inbound_calls_per_operator)
print(total_inbound_calls_per_operator[913942.0])
```

```python
# cantidad total de llamadas entrantes pérdidas por operador

missed_inbound_calls = inbound_calls.groupby('operator_id')['is_missed_call'].sum().reset_index()
print(missed_inbound_calls.sort_values(
    by="is_missed_call", 
    ascending=False))
```

```python
#cambiar indice para poder hacer la división
missed_inbound_calls = missed_inbound_calls.set_index("operator_id")
missed_inbound_calls["missed_ratio"] = ((
    missed_inbound_calls['is_missed_call'] / total_inbound_calls_per_operator
)* 100)

missed_inbound_calls_sorted = missed_inbound_calls.sort_values(
    by="is_missed_call", 
    ascending=False
)

print(missed_inbound_calls_sorted)
```

```python
# 2.Tiempo de espera prolongado para llamadas entrantes
#Aquí si debemos analizar promedio para encontrar deficiencias
dataset_operators['hold'] = dataset_operators['total_call_duration'] - dataset_operators['call_duration']
print(dataset_operators['hold'].describe())
```

```python
total_hold = dataset_operators.groupby('operator_id')['hold'].sum()
total_calls = dataset_operators.groupby('operator_id').size()
avg_hold = (total_hold / total_calls)
avg_hold_df = avg_hold.reset_index()
avg_hold_df.columns = ['operator_id','avg_hold']
print(avg_hold_df)
```

```python
# 3. Llamadas salientes


outbound_calls = dataset_operators[dataset_operators['direction']=='out']
outbound_calls_operators = outbound_calls.groupby('operator_id').size().reset_index()
outbound_calls_operators.columns = ['operator_id','total_outbound_calls']
print(outbound_calls_operators)
```

A continuación junté las métricas en una sola como un df de métricas acumuladas que vi anteriormente en la teoría, sin embargo, sin una separación clara tipo fechas o grupos (como en la mayoría de los callcenter) no es muy fácil encontrar los operadores deficientes en una tabla tan grande como esta, especialmente si vemos que cada operador es deficiente en diferentes métricas. Y como este trabajo solo busca los operadores más deficientes independiente del tiempo, lo mejor es tener una tabla diferente con operadores ineficientes según cada métrica con el tiempo como acumulado.

```python
#métricas individuales : missed_inbound_calls , avg_hold_df y outbound_calls_operators
merged = missed_inbound_calls.merge(avg_hold_df, left_on='operator_id', right_on ='operator_id', how='outer')
merged = merged.merge(outbound_calls_operators, on='operator_id', how='outer')
print(merged.value_counts())
```

```python
#Operadores deficientes respecto a llamadas perdidas
top_missed_calls = missed_inbound_calls_sorted[missed_inbound_calls_sorted['missed_ratio']>6.000000].reset_index()
#Operadores deficientes respecto a tiempo de espera prolongado
top_most_hold = avg_hold_df[avg_hold_df['avg_hold']>300.000000]
#Promedio de llamadas salientes
outbound_calls_stats = outbound_calls_operators.describe()
#Operadores con llamadas por debajo del promedio
low_outbound_operators = outbound_calls_operators[outbound_calls_operators['total_outbound_calls']<35.970000]
print(top_missed_calls)
print(top_most_hold.sort_values(by='avg_hold',ascending=False))
print(outbound_calls_stats)
print(low_outbound_operators)
```
